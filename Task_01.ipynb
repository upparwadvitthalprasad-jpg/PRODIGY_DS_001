```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(
    "/content/API_SP.POP.TOTL_DS2_EN_csv_v2_327505.csv",
    skiprows=4
)

print(df.head())
```

```python
latest_year = "2024"

population = df[["Country Name", latest_year]]
population = population.dropna()

top10 = population.sort_values(
    by=latest_year,
    ascending=False
).head(10)

plt.figure(figsize=(12,6))
sns.barplot(
    x="Country Name",
    y=latest_year,
    data=top10
)

plt.title("Top 10 Most Populated Countries (2024)")
plt.xlabel("Country")
plt.ylabel("Population")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    population[latest_year],
    bins=30,
    kde=True
)

plt.title("Distribution of Population Across Countries")
plt.xlabel("Population")
plt.ylabel("Frequency")
plt.show()
```

```python
india = df[df['Country Name']=='India']

years = ['1960','1970','1980','1990','2000','2010','2020']
values = [india[y].values[0] for y in years]

plt.bar(years, values)
plt.title("India Population Growth")
plt.xlabel("Year")
plt.ylabel("Population")
plt.show()
```

```python
large = population[population['2024'] > 100000000]
large = large.sort_values('2024', ascending=False)

plt.bar(large['Country Name'], large['2024'])
plt.xticks(rotation=90)
plt.show()
```

```python
years = ['1960','1970','1980','1990','2000','2010','2020']
world_pop = [df[y].sum() for y in years]

plt.bar(years, world_pop)
plt.title("World Population Growth by Decade")
plt.xlabel("Year")
plt.ylabel("Population")
plt.show()
```
